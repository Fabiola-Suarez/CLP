## 5. Limpieza de datos

Aunque el dataset ya fue creado de forma ordenada, se aplica una limpieza basica para respetar el proceso de analisis.

```python
df_limpio = df.drop_duplicates().copy()
df_limpio = df_limpio.fillna("N/A")
df_limpio["lenguaje"] = df_limpio["lenguaje"].str.strip()
df_limpio["paradigma_principal"] = df_limpio["paradigma_principal"].str.lower().str.strip()

df_limpio.head()
```

## 6. Top 10 de lenguajes mas usados

Se toman los diez primeros puestos del ranking para ver cuales son los lenguajes mas populares.

```python
top_10 = df_limpio.sort_values("ranking_2026").head(10)
top_10
```

```python
plt.figure(figsize=(10, 5))
plt.bar(top_10["lenguaje"], top_10["porcentaje"], color="#2f6f8f")
plt.title("Top 10 de lenguajes mas usados")
plt.xlabel("Lenguaje")
plt.ylabel("Popularidad (%)")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 7. Comparacion entre ranking 2026 y ranking 2025

La diferencia de ranking permite ver que lenguajes subieron, bajaron o se mantuvieron.

```python
df_limpio["cambio_posicion"] = df_limpio["ranking_2025"] - df_limpio["ranking_2026"]

df_limpio[["lenguaje", "ranking_2026", "ranking_2025", "cambio_posicion"]].sort_values(
    "cambio_posicion", ascending=False
)
```

```python
cambios = df_limpio.sort_values("cambio_posicion", ascending=True)
colores = ["#b56576" if valor < 0 else "#4c956c" if valor > 0 else "#6c757d" for valor in cambios["cambio_posicion"]]

plt.figure(figsize=(10, 6))
plt.barh(cambios["lenguaje"], cambios["cambio_posicion"], color=colores)
plt.title("Cambio de posicion respecto al ano anterior")
plt.xlabel("Posiciones ganadas o perdidas")
plt.ylabel("Lenguaje")
plt.axvline(0, color="black", linewidth=0.8)
plt.grid(axis="x", alpha=0.3)
plt.show()
```

## 8. Lenguajes que mas crecieron y mas bajaron

```python
mas_crecieron = df_limpio.sort_values("variacion", ascending=False).head(5)
mas_bajaron = df_limpio.sort_values("variacion").head(5)

display(mas_crecieron[["lenguaje", "porcentaje", "variacion"]])
display(mas_bajaron[["lenguaje", "porcentaje", "variacion"]])
```

```python
plt.figure(figsize=(10, 5))
plt.bar(df_limpio["lenguaje"], df_limpio["variacion"], color="#6a994e")
plt.title("Variacion de popularidad en puntos porcentuales")
plt.xlabel("Lenguaje")
plt.ylabel("Variacion")
plt.xticks(rotation=60, ha="right")
plt.axhline(0, color="black", linewidth=0.8)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 9. Paradigmas principales

Se agrupan los lenguajes por el paradigma principal asignado en el dataset.

```python
conteo_paradigmas = df_limpio["paradigma_principal"].value_counts()
conteo_paradigmas
```

```python
plt.figure(figsize=(8, 5))
conteo_paradigmas.sort_values().plot(kind="barh", color="#b56576")
plt.title("Cantidad de lenguajes por paradigma principal")
plt.xlabel("Cantidad")
plt.ylabel("Paradigma")
plt.grid(axis="x", alpha=0.3)
plt.show()
```

## 10. Conclusion

El ranking muestra que Python ocupa el primer lugar entre los lenguajes mas populares, seguido por C, C++, Java y C#.
